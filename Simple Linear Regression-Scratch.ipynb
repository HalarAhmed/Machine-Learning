```python
class LR: # Linear Regression Class
  def __init__(self):
    self.m=None
    self.b=None
  def fit(self,x_train,y_train):  # function to train/find best 'm' and 'b'
    num=0
    den=0
    for i in range(x_train.shape[0]):
      num=num+((x_train[i]-x_train.mean())*(y_train[i]-y_train.mean()))
      den=den+((x_train[i]-x_train.mean())*(x_train[i]-x_train.mean()))

    self.m=num/den
    self.b=y_train.mean()-(self.m*x_train.mean())
    print(self.m)
    print(self.b)

  def predict(self,x_test): # function to predict using testing datset
    return (self.m*(x_test))+self.b
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("placement.csv")
df.head()
```

```python
x=df.iloc[:,0].values # input as an array
y=df.iloc[:,1].values # output as an array
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
x_train.shape
```

```python
lr=LR()
```

```python
lr.fit(x_train,y_train)
```

```python
print(lr.predict(x_test[0]))
```
